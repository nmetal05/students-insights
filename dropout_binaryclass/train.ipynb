```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, RobustScaler
```

```python
model = Pipeline([
    ('scaler', StandardScaler()),
    ('clf', LogisticRegression(
    penalty='l2',
    C=1.0,
    solver='lbfgs',
    class_weight='balanced',
    random_state=42
))
])
```

```python
# First, ensure Target is properly transformed
print("Checking Target before processing:")
print(f"dtype: {y.dtype}")
print(f"unique values: {y.unique()}")
print(f"null count: {y.isnull().sum()}")

# Fix the target variable
y = df['Target'].map({'Dropout': 0, 'Graduate': 1})

# Handle any NaN values that might have appeared
if y.isnull().any():
    print(f"Found {y.isnull().sum()} null values, dropping them...")
    mask = y.notnull()
    x = x[mask]
    y = y[mask]

# Convert to integer
y = y.astype(int)

# Verify the fix
from sklearn.utils.multiclass import type_of_target
print(f"\nAfter fix:")
print(f"dtype: {y.dtype}")
print(f"unique values: {y.unique()}")
print(f"Target type: {type_of_target(y)}")

# Now run the cross-validation
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
auc_roc_scores = []
acc_scores = []

for fold, (train_index, val_index) in enumerate(skf.split(x, y), 1):
    x_train, x_val = x.iloc[train_index], x.iloc[val_index]
    y_train, y_val = y.iloc[train_index], y.iloc[val_index]

    model.fit(x_train, y_train)

    y_pred = model.predict(x_val)
    y_pred_proba = model.predict_proba(x_val)[:, 1]

    auc_roc = roc_auc_score(y_val, y_pred_proba)
    acc = accuracy_score(y_val, y_pred)

    auc_roc_scores.append(auc_roc)
    acc_scores.append(acc)

    print(f"fold: {fold}: \n")
    print(f"accuracy: {acc}, roc_auc: {auc_roc}")
    print("classification report: ")
    print(classification_report(y_val, y_pred, target_names=['Dropout', 'Graduate']))
    print("----------------------------------------------- \n")

print(f"average roc_auc: {np.mean(auc_roc_scores):.4f} +- {np.std(auc_roc_scores):.4f}")
print(f"average accuracy: {np.mean(acc_scores):.4f} +- {np.std(acc_scores):.4f}")
```

```python
print(f"average roc_auc: {np.mean(auc_roc_scores)} +- {np.std(auc_roc_scores):.4f}")
print(f"average accuracy: {np.mean(acc_scores)} +- {np.std(acc_scores):.4f}")
```

```python
## now train the model with full data after CV
final_model = model.fit(x,y)
```
